```python
! pip install Faker
from faker import Faker
import pandas as pd
import pyarrow as pa
import pyarrow.parquet as pq

faker = Faker("sk_SK")

def get_users_data(count):
    users= []

    for i in range(1,count + 1):
        name = faker.first_name()
        surname = faker.last_name()
        email = f"{name.lower()}.{surname.lower()}{i}@gmail.com"
        adress = faker.address()
        phone = faker.phone_number()

        user = {
            "number" :i,
            "name": name,
            "surname" : surname,
            "email" : email,
            "adress" : adress,
            "phone" : phone
        }

        users.append(user)
    return users

users = get_users_data(500)

for user in users:
    print(f"{user['number']}. user: {user['name']} {user['surname']}, {user['email']}")

df_users = pd.DataFrame(users)
df_users.to_csv("export_users.csv", index= False, encoding = "utf-8")

table = pa.Table.from_pandas(df_users)
pq.write_table(table, 'export-users.parquet')

```

```python
!pip install openpyxl
import polars as pl
import pandas
import pyarrow as pa
import pyarrow.parquet as pq
import pyarrow.orc as orc
import pyarrow.feather as feather
import openpyxl

# Dataset with online courses
courses = pl.DataFrame({
    "course_name": [
        "Python Základy", "Dátová Analýza", "HTML, CSS Web Vývoj", "Pokročilý Python", "Strojové Učenie", "AI Základy", "Microsoft Excel", "Java Základy", "C# Programovanie", "Docker", "Cloud Computing", "Data Engineering", "Dátová Vizuálizácia", "UX/UI Design", "JavaScript Základy", "Správa Serverov", "Hacking a Bezpečnosť", "Deep Learning", "Testovanie Softvéru", "DevOps"],
    "price": [150, 180, 200, 220, 250, 270, 190, 160, 300, 250, 
             220, 230, 270, 180, 210, 250, 200, 240, 270, 290],
    "Category": [
        "Python", "Dátová Analýza", "Web Development", "Python", "AI", "AI", "Microsoft Excel", "Java", "C#", "Cloud Computing", "Cloud Computing", "Data Engineering", "Dátová Analýza", "UX/UI Dizajn", "Web Vývoj", "DevOps", "Bezpečnosť", "AI", "Testovanie", "DevOps"],
    "course_length": [10, 20, 30, 15, 25, 18, 22, 28, 24, 21, 
                    27, 19, 23, 30, 16, 25, 21, 20, 18, 26]
})

print(courses)
```

```python
# Filtering courses with a price above €150
courses_filtered = courses.filter(pl.col("price") > 150)
print(courses_filtered)
```

```python
# Filtering courses that are longer than 20 hours
courses_filtered = courses.filter(pl.col("course_length") > 20)
print(courses_filtered)
```

```python
# Filtering the longest course
longest_course = courses.sort("course_length", descending=True).head(1)
print(longest_course)
     
```

```python
# Filter by category and price
courses_python_expensive = courses.filter(
    (pl.col("Category") == "Python") & (pl.col("price") > 200)
)
print(courses_python_expensive)
     
```

```python
# Aggregation - average price by category
agg_1 = courses.group_by("Category").agg(pl.col("price").mean().alias("mean_price"))
print(agg_1)
     
```

```python

# Number of courses in each category
count_courses_category = courses.group_by("Category").agg(pl.col("course_name").count().alias("count_courses"))
print(count_courses_category)
     
```

```python
# Number of courses in each category, sorted from largest to smallest
count_courses_category = courses.group_by("Category").agg(pl.col("course_name").count().alias("count_courses"))

# Sort by number of courses from largest to smallest
count_courses_category_sorted = count_courses_category.sort("count_courses", descending=True)


print(count_courses_category_sorted)
```

```python
# Summary - total price of all courses
total_price = courses.select(pl.col("price").sum().alias("total_price"))
print(total_price)
```

```python
# Summary: total price and average course length
summary = courses.select(
    pl.col("price").sum().alias("total_price"),
    pl.col("course_length").mean().alias("mean_length")
)
print(summary)
```

```python

# Adding a "new" column
courses = courses.with_columns(
    (pl.col("price") > 200).alias("new")
)
print(courses)
     
```

```python

# Maximum course length
max_length_course = courses.select(pl.col("course_length").max().alias("max_length"))
print(max_length_course)
```

```python
import polars as pl
import random
import numpy as np
from datetime import datetime, timedelta


RECORD_COUNT = 500_000


data_clients = {
   
    'id': [i for i in range(1, RECORD_COUNT + 1)],  
    "name": np.random.choice([
        "Adriana", "Alena", "Andrea", "Anna", "Barbora", "Beáta", "Daniela", "Dominika", "Elena", "Emília", "Erika", "Eva", "Gabriela", "Hana", "Ivana", "Jana", "Katarína", "Kristína", "Laura", "Lenka", "Lucia", "Magdaléna", "Mária", "Martina", "Monika", "Natália", "Petra", "Renáta", "Silvia", "Simona", "Soňa", "Tamara", "Veronika", "Zuzana"], size=RECORD_COUNT),
    'age': [random.randint(21, 71) for _ in range(1, RECORD_COUNT + 1)],  
    'purchase': [random.uniform(150, 300) for _ in range(1, RECORD_COUNT + 1)],  
    'country': [random.choice(['Slovensko', 'Cesko', 'Polsko']) for _ in range(1, RECORD_COUNT + 1)], 
    'date': [datetime(2024, 1, 1) + timedelta(days=random.randint(0, 365)) for _ in range(1, RECORD_COUNT + 1)], 
}

df_clients = pl.DataFrame(data_clients)


print(df_clients.head())

mean_age = df_clients.select(pl.col('age').mean())
mean_purchase = df_clients.select(pl.col('purchase').mean())
mean_date = df_clients.select(pl.col('date').mean())

print(f'Priemerny vek: {mean_age}')
print(f'Priemerny nakupy: {mean_purchase}')
print(f'Priemerny datum nakupu: {mean_date}')
```
