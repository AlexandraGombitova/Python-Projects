## Property

```python
class Product:
    def __init__(self, price):
        self.price = price

    @property
    def price(self):
        return self.price
    
    @price.setter
    def price(self, value):
        if value < 0:
            raise ValueError("Price can not be lover than 0")
        
```

```python
class Person:
    def __init__(self):
        self.__age = 0
    
    @property
    def age(self):
        return self.__age
    
    @age.setter
    def age(self, value):
        if value  < 0:
            raise ValueError("Age cannot be less than 0")
        self.__age = value


carol = Person()
carol.age = 60
print(f"Person is: {carol.age} years old")
```

## Duck Typing

```python
def movement(object):
    object.move()

class Duck:
        def move(self):
            print("DUCK MOVE")

class Person:
    def move(self):
        print("PERSON MOVE")
        

duck_donald = Duck()
person_alex = Person()

movement(duck_donald)
movement(person_alex)

        
```

```python
def play_instrument(instrument):
    instrument.play_sound()

class Guitar:
    def play_sound(self):
        print("Guitar - sound")

class Piano:
    def play_sound(self):
        print("Piano - sound")

class Drums:
    def play_sound(self):
        print("Drums - sound")

guitar = Guitar()
piano = Piano()
drums = Drums()

play_instrument(guitar)
play_instrument(piano)
play_instrument(drums)
```

## Cache

```python
import math

class Circle:
    def __init__(self, radius):
        self.radius = radius
        self._area = None

    @property
    def area(self):
        if self._area is None:
            print("Counting the area")
            self._area = math.pi * math.pow(self.radius, 2)
        return self._area

circle = Circle(10)
print(circle.area)
        
```

## Static method

```python
class Calculator:
    
    @staticmethod
    def count(a, b):
        return a + b



print(Calculator.count(2,3))
```

```python
import math
class MathematicOperaions:
    @staticmethod
    def area_circle(radius):
        return math.pi * (radius ** 2)
    
    @staticmethod
    def area_square(a):
        return a ** 2
    
    
print(MathematicOperaions.area_circle(10))
```

```python
class Temperature:
    @staticmethod
    def celsius_to_fahrenheit(celsius):
        return(celsius * 9/5) + 32
    
    @staticmethod
    def fahrenheit_to_celsius(fahrenheit):
        return(fahrenheit - 32) * 5/9
    

celsius = 30
fahrenheit = Temperature.celsius_to_fahrenheit(celsius)
print(f"{celsius} celsius je {fahrenheit} fahrenheit")
```
