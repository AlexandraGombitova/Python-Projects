```python
from keras.datasets import imdb

(x_train, y_train), (x_test, y_test) = imdb.load_data(num_words= 10000)


word_index = imdb.get_word_index()

word_index = {k: (v + 3) for k, v in word_index.items()}
word_index["<PAD>"] = 0
word_index["<START>"] = 1
word_index["<UNK>"] = 2
word_index["<UNUSED>"] = 3


reverse_word_index = {value: key for key,value in word_index.items()}

def decode_review(text_indexes):
    return" ".join([reverse_word_index.get(i, "?")for i in text_indexes])


print(decode_review(x_train[1]))
```

```python
import matplotlib.pyplot as plt
import numpy as np
from keras.models import Sequential
from keras.layers import Dense

model = Sequential()
model.add(Dense(64, activation='relu', input_shape=(10,)))
model.add(Dense(1, activation='sigmoid'))

model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])


x_train = np.random.rand(1000, 10)
y_train = np.random.randint(0, 2, size=(1000, 1))


history = model.fit(x_train, y_train, epochs=10, batch_size=32, verbose=0)

loss_values = history.history['loss']
acc_values = history.history['accuracy']
epochs = range(1, len(loss_values) + 1)


plt.plot(epochs, loss_values, 'bo-', label='Loss (loss)')
plt.plot(epochs, acc_values, 'go-', label='Accuracy (accuracy)')
plt.title('Journey of train model')
plt.xlabel('Epocha')
plt.ylabel('Value')
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from keras.models import Sequential
from keras.layers import Dense


model = Sequential()
model.add(Dense(64, activation='relu', input_shape=(10,)))
model.add(Dense(1, activation='sigmoid'))

model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])


x_data = np.random.rand(1000, 10)
y_data = np.random.randint(0, 2, size=(1000, 1))


history = model.fit(
    x_data,
    y_data,
    epochs=10,
    batch_size=32,
    validation_split=0.2,
    verbose=0
)


epochs = range(1, len(history.history['loss']) + 1)

plt.plot(epochs, history.history['loss'], 'bo-', label='Tréningová strata')
plt.plot(epochs, history.history['val_loss'], 'b--', label='Validačná strata')
plt.plot(epochs, history.history['accuracy'], 'go-', label='Tréningová presnosť')
plt.plot(epochs, history.history['val_accuracy'], 'g--', label='Validačná presnosť')

plt.title('Journey of train and validation of model')
plt.xlabel('Epocha')
plt.ylabel('Value')
plt.legend()
plt.grid(True)
plt.show()
```

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from keras.layers import Dense
from keras.models import Sequential


(x_train, y_train), (x_test, y_test) = keras.datasets.mnist.load_data()


x_train = x_train.reshape(-1, 28*28).astype("float32") / 255
x_test = x_test.reshape(-1, 28*28).astype("float32") / 255


model = keras.Sequential([
    layers.Dense(128, activation='relu', input_shape=(784,)),
    layers.Dense(10, activation='softmax')
])


model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)


model.fit(x_train, y_train, epochs=5, batch_size=32)


test_loss, test_acc = model.evaluate(x_test, y_test)
print(f"Accuracy on test data: {test_acc:.2f}")

predictions = model.predict(x_test[:5])
print("Predictions:", predictions.argmax(axis=1))
print("Real Values:", y_test[:5])
```

```python

import tensorflow as tf
from tensorflow import keras
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten
from tensorflow.keras.datasets import mnist
from tensorflow.keras.utils import to_categorical


(X_train, y_train), (X_test, y_test) = mnist.load_data()

X_train = X_train / 255.0
X_test = X_test / 255.0

y_train = to_categorical(y_train, 10)
y_test = to_categorical(y_test, 10)

print(f"✅ Dataset load. Count of train examples: {X_train.shape[0]}")
print(f"📐 Size of each picture: {X_train.shape[1:]} (28x28)")
print(f"🧾 Category example (one-hot): {y_train[0]}")


model = Sequential([
    Flatten(input_shape=(28, 28)),      
    Dense(128, activation='relu'),       
    Dense(10, activation='softmax')      
])

print(model)

model.summary()



model.compile(
    optimizer='adam',                         
    loss='categorical_crossentropy',           
    metrics=['accuracy']                       
)

print("⚙️ Model was compiled.")
print("Optimizer used: Adam")
print("Loss category: categorical_crossentropy")
print("Metrics: accuracy")


history = model.fit(
    X_train, y_train,
    epochs=5,              
    batch_size=32,         
    validation_split=0.1   
)

print(history)

print("✅ Training finnished.")
print(f"📈 Last accuracy: {history.history['accuracy'][-1]:.4f}")
print(f"📉 Last validation: {history.history['val_accuracy'][-1]:.4f}")


test_loss, test_accuracy = model.evaluate(X_test, y_test)


print(f"🧪 Loss on test data: {test_loss:.4f}")
print(f"🎯 Accuracy on test data: {test_accuracy:.4f}")


import matplotlib.pyplot as plt


plt.plot(history.history['accuracy'], label='Training accuracy')
plt.plot(history.history['val_accuracy'], label='Validation accuracy')
plt.title('Accuracy of model during training')
plt.xlabel('Epocha')
plt.ylabel('Accuracy')
plt.legend()
plt.grid(True)
plt.show()

plt.plot(history.history['loss'], label='Training Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.title('Model loss during training')
plt.xlabel('Epocha')
plt.ylabel('Loss')
plt.legend()
plt.grid(True)
plt.show()
```
