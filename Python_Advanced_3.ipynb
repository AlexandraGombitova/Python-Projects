## @Classmethod

```python
class Employee:
    employee_count = 0

    def __init__(self, name, salary):
        self.name = name
        self.salary = salary
        Employee.employee_count += 1
    
    @classmethod
    def get_db(cls, record):
        name, salary = record.split(",")
        return cls(name, int(salary))

employee_record = "Alex Madagascar, 7000"
employee_record2 = "Momo Madagascar, 7000"
new_employee = Employee.get_db(employee_record)
new_employee2 = Employee.get_db(employee_record2)

        
```

```python
import xml.etree.ElementTree as ET


xml_data = """<?xml version="1.0"?>
<data>
<person>
    <name>Alex</name>
    <surname>Mort</surname>
    <age>88</age>
    <city>Bratislava</city>
</person>
</data>
"""
root = ET.fromstring(xml_data)

for person in root.findall("person"):
    name = person.find("name").text
    print(f"Name: {name}")
```

## Clustering

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
import numpy as np

X = np.random.rand(100, 2)
kmeans = KMeans(n_clusters = 3, n_init = 10)
kmeans.fit(X)
y_means = kmeans.predict(X)

plt.scatter(X[:,0], X[:,1],c = y_means, s = 50, cmap="viridis")
centers = kmeans.cluster_centers_
plt.scatter(centers[:, 0], centers[:, 1], c = "red", s=400, alpha = 0.7)
plt.show()
```

```python
# Load required libraries and create sample dataset
import pandas as pd
import numpy as np

# Create sample DataFrame with missing values
df = pd.DataFrame({
    'A': [1, np.nan, 3, np.nan, 5],
    'B': [np.nan, 2, 3, 4, 5],
    'C': [1, 2, np.nan, 4, 5]
})

# One-liner to handle missing values with multiple strategies
cleaned_df = df.fillna(df.mean()).where(df.notnull(), df.fillna(method='ffill'))

# Display results
print("Original DataFrame:")
print(df)
print("\nCleaned DataFrame:")
print(cleaned_df)
```

```python
import pandas as pd
import random 

names_example = ["Iva", "Mort", "Julien", "Peter", "Alex", "Saska", "Moris"]
surnames_example = ["Deep", "Ford", "Cook", "Depalo", "McCalen", "Gruber", "Wohner"]

data = {
    "ID" : [i for i in range(1, 501)],
    "Name" : [random.choice(names_example) for _ in range(500)],
    "Surname" : [random.choice(surnames_example) for _ in range(500)],
    "Age" : [random.randint(21, 75) for _ in range(500)],
}

df_customers = pd.DataFrame(data)
df_customers
df_customers.to_csv("customers.csv", index = False, encoding = "utf-8")
```

```python
import numpy as np
import pandas as pd
from sklearn.impute import SimpleImputer

data = pd.DataFrame({
    "Name": ["Momom", "Kate", "Peter", None, "Teddy"],
    "Age" : [25, 30, np.nan, 27, 35],
    "Values" : [5, np.nan, 3, 8, 10]
})

imputer = SimpleImputer(strategy = "mean")
data[["Age", "Values"]] = imputer.fit_transform(data[["Age", "Values"]])

print(data)
```

```python
import numpy as np
import pandas as pd
from sklearn.impute import SimpleImputer
from sklearn.impute import KNNImputer


data = pd.DataFrame({
    "Name": ["Momom", "Kate", "Peter", None, "Teddy"],
    "Age" : [25, 30, np.nan, 27, 35],
    "Values" : [5, np.nan, 3, 8, 10]
})


knn_imputer = KNNImputer(n_neighbors=2)
data[["Age", "Values"]] = knn_imputer.fit_transform(data[["Age", "Values"]])

print(data)
```
